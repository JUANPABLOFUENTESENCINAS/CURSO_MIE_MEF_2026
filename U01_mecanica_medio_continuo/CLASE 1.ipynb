### Sentencia IF

```octave
%Coeficientes del polinomio de segundo grado
a=1;
b=7;
c=1;
% Calculo el discriminante
disc = b ^2 - 4* a*c

%Bloque de condicion 
if disc > 0
    x1 = (-b +sqrt( disc )) / (2* a);
    x2 = (-b -sqrt( disc )) / (2* a);
    disp(x1)
    disp(x2)
    disp('Dos raices reales distintas')

    elseif disc == 0
        x1 = -b / (2* a);
        disp(x1)
        disp('Una raiz real multiple')
    else
        disp('Raices complejas')
end
```

### Sentencia FOR
$$ \sum_{n=1}^{\N} \frac{1}{n^2} = \frac{\pi^2}{6} $$


```octave
n=100000;
suma=0;
%valor de referencia
Sum_exac=pi^2/6

for i=1:n
suma=suma+1/i^2;
end
suma
error=(abs(Sum_exac-suma)/Sum_exac)*100
```

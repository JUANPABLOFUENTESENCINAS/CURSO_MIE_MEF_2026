```octave
n=1000;
suma=0; 

%valor de referencia
Sum_exac=pi^2/6

for i=1:n
    suma=suma+1/i^2;
    error=abs(suma-Sum_exac)/Sum_exac*100;
    Lista(i, :) = [i, suma, error];
end
% Encabezados
fprintf('%-8s %-18s %-15s\n', 'n', 'Suma_Acumulada', 'Error');

% Datos (se transpone Lista con ' porque fprintf lee por columnas)
fprintf('%-8d %-18.6f %-15.6f\n', Lista');
```

### Sentencia While

$$\sum_{n=1}^{\infty} \frac{1}{n^2} = \frac{\pi^2}{6}$$

```octave
% --- 1. CONFIGURACIÓN DE LA COTA DE ERROR ---
tolerancia = 1e-2;          % Cota de error permitida (ej: 0.0001)
valor_teorico = pi^2 / 6;   
format short
suma_acumulada = 0;
n = 1;
error_calculado = valor_teorico; 

Lista=zeros(0,3); % Inicializamos la lista para almacenar los resultados

while error_calculado > tolerancia
    suma_acumulada = suma_acumulada + 1 / (n^2);
      
    error_calculado = abs(valor_teorico - suma_acumulada);
    
    List(n, :) = [n, suma_acumulada, error_calculado];

    % Incrementamos el contador para el siguiente término
    n = n + 1;
end

% Encabezados
fprintf('%-8s %-18s %-15s\n', 'n', 'Suma_Acumulada', 'Error');

% Datos (se transpone Lista con ' porque fprintf lee por columnas)
fprintf('%-8d %-18.6f %-15.6f\n', List');
```

### Implementacion manual del algoritmo de Cholesky

La funcion `chol()` resuelve todo de una vez, pero el metodo se puede programar elemento a elemento con bucles `for` (algoritmo de Cholesky-Banachiewicz), igual que hicimos antes con la suma de la serie:

$$L_{jj}=\sqrt{A_{jj}-\sum_{k=1}^{j-1}L_{jk}^2} \qquad\qquad L_{ij}=\frac{1}{L_{jj}}\left(A_{ij}-\sum_{k=1}^{j-1}L_{ik}L_{jk}\right),\ \ i>j$$

Solo se recorre el triangulo inferior de A porque, al se

```octave
%Implementacion manual de Cholesky (algoritmo de Cholesky-Banachiewicz)
A=[2 -1 0;
   -1 2 -1;
    0 -1 1]  %matriz simetrica definida positiva (ejemplo clasico)

n=size(A,1);
L=zeros(n);

for i=1:n
    for j=1:i
        suma=0;
        for k=1:j-1
            suma=suma+L(i,k)*L(j,k);
        end
        if i==j
            L(i,j)=sqrt(A(i,i)-suma);
        else
            L(i,j)=(A(i,j)-suma)/L(j,j);
        end
    end
end

L  %factor triangular inferior calculado manualmente

%Verificacion contra la funcion nativa de Octave
L_octave=chol(A)'  %chol() devuelve el triangular superior U, por eso se transpone

error_max=max(max(abs(L-L_octave)))  %debe ser practicamente cer
```

### IMPLEMENTACION ALGORITMO DE INTEGRACION TRAPECIO

```octave
% Ejemplo: integral de sin(x) entre 0 y pi (valor exacto = 2)

f = @(x) sin(x);   % funcion a integrar
a = 0;             % limite inferior
b = pi;            % limite superior
n = 100;           % numero de subintervalos

h = (b - a) / n;

% Calculamos los puntos x(i) con un for
x = zeros(1, n+1);
for i = 1:n+1
    x(i) = a + (i-1)*h;
end

% Evaluamos la funcion en cada punto con un for
y = zeros(1, n+1);
for i = 1:n+1
    y(i) = f(x(i));
end

% Sumamos los terminos intermedios con un for
suma = 0;
for i = 2:n
    suma = suma + y(i);
end

I = (h/2) * (y(1) + 2*suma + y(n+1));

printf("Integral aproximada (trapecio): %f\n", I);
printf("Valor exacto: %f\n", 2);
printf("Error: %e\n", abs(I - 2));
```
